## Attempt 3: Dual Input Audio and Text based Recurrent model

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader, Subset
import numpy as np
import pandas as pd
import os
import nltk
from collections import Counter
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, f1_score, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
from tqdm import tqdm
try:
    from IPython.display import Audio, display
    IPYTHON_AVAILABLE = True
except ImportError:
    IPYTHON_AVAILABLE = False

# Configuration
DATA_DIR = "/kaggle/input/"
CSV_PATH = os.path.join(DATA_DIR, "/kaggle/working/iemocap_regression_metadata.csv")
DATASET_DIR = os.path.join(DATA_DIR, "iemocapfullrelease/IEMOCAP_full_release")
MFCC_PATH = os.path.join(DATA_DIR, "mfcc-regression/")
MODEL_SAVE_PATH = "/kaggle/working/multimodal_emotion_classifier.pt"
LOSS_CURVE_PATH = "/kaggle/working/multimodal_loss_curve.png"
F1_CURVE_PATH = "/kaggle/working/multimodal_f1_curve.png"
GLOVE_PATH = "/kaggle/working/glove.6B.300d.txt"
BATCH_SIZE = 16
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
EMOTION_CLASSES = ["angry", "happy", "sad", "neutral"]
LABEL_MAPPING = {"ang": "angry", "hap": "happy", "sad": "sad", "neu": "neutral"}
MAX_LENGTH = 128
EMBEDDING_DIM = 300
HIDDEN_DIM = 256
NUM_LAYERS = 2
DROPOUT = 0.3
N_MFCC = 13
MAX_AUDIO_LENGTH = 300
EPOCHS = 30
LEARNING_RATE = 0.001
SESSIONS = ['Session1', 'Session2', 'Session3', 'Session4', 'Session5']

nltk.download('punkt')

# Placeholder for GloVe embeddings
# Load GloVe embeddings
def load_glove_embeddings():
    embeddings_index = {}
    try:
        with open(GLOVE_PATH, 'r', encoding='utf-8') as f:
            for line in tqdm(f, desc="Loading GloVe embeddings"):
                values = line.strip().split()
                word = values[0]
                try:
                    coefs = np.asarray(values[1:], dtype='float32')
                    if coefs.shape[0] == EMBEDDING_DIM:
                        embeddings_index[word] = coefs
                except ValueError:
                    continue
        print(f"Loaded {len(embeddings_index)} GloVe embeddings")
    except FileNotFoundError:
        print(f"Error: GloVe file not found at {GLOVE_PATH}")
        raise
    except Exception as e:
        print(f"Error loading GloVe embeddings: {e}")
        raise
    return embeddings_index

# Multimodal Dataset
class Multimodal_IEMOCAPDataset(Dataset):
    def __init__(self, csv_path, dataset_dir, mfcc_path, sessions, max_length=128, n_mfcc=13, max_audio_length=300):
        self.data = pd.read_csv(csv_path)
        self.dataset_dir = dataset_dir
        self.mfcc_path = mfcc_path
        self.max_length = max_length
        self.n_mfcc = n_mfcc
        self.max_audio_length = max_audio_length
        self.sessions = sessions
        
        # Emotion mapping
        self.emotion_to_idx = {"angry": 0, "happy": 1, "sad": 2, "neutral": 3}
        self.data['mapped_emotion'] = self.data['emotion'].str.lower().map(LABEL_MAPPING)
        self.data = self.data[self.data['mapped_emotion'].isin(EMOTION_CLASSES)]
        
        # Load transcripts
        self.transcripts = {}
        for session in sessions:
            trans_dir = os.path.join(dataset_dir, session, 'dialog', 'transcriptions')
            if os.path.exists(trans_dir):
                for trans_file in os.listdir(trans_dir):
                    if trans_file.endswith('.txt'):
                        with open(os.path.join(trans_dir, trans_file), 'r') as f:
                            for line in f:
                                if line.strip():
                                    utt_id = line.split('[')[0].strip()
                                    text = line.split(':')[-1].strip()
                                    self.transcripts[utt_id] = text
        print(f"Number of transcripts loaded: {len(self.transcripts)}")
        
        # Filter data for valid transcripts and MFCCs
        self.data = self.data[self.data['utt_id'].isin(self.transcripts.keys())]
        self.data = self.data[self.data['mfcc_path'].apply(lambda x: os.path.exists(os.path.join(mfcc_path, x)))]
        print(f"Filtered data size: {len(self.data)}")
        
        if len(self.data) == 0:
            raise ValueError("Dataset is empty after filtering.")
        
        # Compute MFCC statistics
        self.mean, self.std = self.compute_statistics(csv_path)
        
        # Build vocabulary
        self.word2idx = {'<PAD>': 0, '<UNK>': 1}
        word_counts = Counter()
        for text in self.transcripts.values():
            tokens = nltk.word_tokenize(text.lower())
            word_counts.update(tokens)
        for word, count in word_counts.items():
            if count >= 3:
                self.word2idx[word] = len(self.word2idx)
        
        # Load GloVe embeddings
        embeddings_index = load_glove_embeddings()
        embedding_matrix = np.zeros((len(self.word2idx), EMBEDDING_DIM))
        for word, idx in self.word2idx.items():
            if word in embeddings_index:
                embedding_matrix[idx] = embeddings_index[word]
            else:
                embedding_matrix[idx] = np.random.normal(0, 0.1, EMBEDDING_DIM)
        self.embedding_matrix = torch.tensor(embedding_matrix, dtype=torch.float32)
    
    def compute_statistics(self, csv_path):
        data = pd.read_csv(csv_path)
        mfccs = []
        for _, row in data.iterrows():
            try:
                mfcc = np.load(os.path.join(self.mfcc_path, row['mfcc_path']))
                mfccs.append(mfcc)
            except FileNotFoundError:
                continue
        if not mfccs:
            raise ValueError("No valid MFCC files found")
        mfccs = np.concatenate(mfccs, axis=0)
        mean = torch.tensor(mfccs.mean(axis=0), dtype=torch.float32)
        std = torch.tensor(mfccs.std(axis=0), dtype=torch.float32)
        return mean, std
    
    def __len__(self):
        return len(self.data)
    
    def __getitem__(self, idx):
        row = self.data.iloc[idx]
        utt_id = row['utt_id']
        session = row['session']
        dialogue = utt_id.rsplit('_', 1)[0]
        
        # Load MFCC
        try:
            mfcc = np.load(os.path.join(self.mfcc_path, row['mfcc_path']))
        except FileNotFoundError:
            print(f"Error: MFCC file not found at {row['mfcc_path']}")
            raise
        mfcc = torch.tensor(mfcc, dtype=torch.float32)
        
        # Normalize MFCCs
        mfcc = (mfcc - self.mean) / (self.std + 1e-8)
        
        # Compute sequence length
        seq_len = min(mfcc.shape[0], self.max_audio_length)
        
        # Pad or truncate MFCC
        if mfcc.shape[0] < self.max_audio_length:
            pad = torch.zeros((self.max_audio_length - mfcc.shape[0], mfcc.shape[1]))
            mfcc = torch.cat([mfcc, pad], dim=0)
        else:
            mfcc = mfcc[:self.max_audio_length]
        
        # Get context and target text
        dialogue_data = self.data[self.data['utt_id'].str.startswith(dialogue)]
        dialogue_data = dialogue_data.sort_values('utt_id')
        idx_in_dialogue = dialogue_data.index.get_loc(row.name)
        context_utts = dialogue_data.iloc[max(0, idx_in_dialogue-5):idx_in_dialogue]['utt_id'].tolist()
        context_texts = [self.transcripts.get(utt, "") for utt in context_utts]
        target_text = self.transcripts.get(utt_id, "")
        
        # Tokenize and encode text
        context_text = " ".join(context_texts)
        full_text = context_text + " " + target_text
        tokens = nltk.word_tokenize(full_text.lower())[:self.max_length]
        input_ids = [self.word2idx.get(token, self.word2idx['<UNK>']) for token in tokens]
        if len(input_ids) < self.max_length:
            input_ids += [self.word2idx['<PAD>']] * (self.max_length - len(input_ids))
        input_ids = torch.tensor(input_ids, dtype=torch.long)
        
        # Target
        mapped_emotion = row['mapped_emotion']
        target = torch.tensor(self.emotion_to_idx[mapped_emotion], dtype=torch.long)
        
        return input_ids, mfcc, target, utt_id, context_texts, target_text, seq_len

# Collate function
def multimodal_collate_fn(batch):
    input_ids, mfccs, targets, utt_ids, context_texts, target_texts, seq_lens = zip(*batch)
    input_ids = torch.stack(input_ids)
    mfccs = torch.stack(mfccs)
    targets = torch.stack(targets)
    seq_lens = torch.tensor(seq_lens, dtype=torch.long)
    return input_ids, mfccs, targets, list(utt_ids), list(context_texts), list(target_texts), seq_lens

class Multimodal_LSTMEmotionClassifier(nn.Module):
    def __init__(self, vocab_size, embedding_dim, hidden_dim, num_layers, num_classes, dropout, embedding_matrix, n_mfcc=13):
        super(Multimodal_LSTMEmotionClassifier, self).__init__()
        
        # Text branch (BiLSTM)
        self.embedding = nn.Embedding.from_pretrained(embedding_matrix, freeze=False)
        self.text_bilstm = nn.LSTM(
            input_size=embedding_dim,
            hidden_size=hidden_dim,
            num_layers=num_layers,
            batch_first=True,
            bidirectional=True,
            dropout=dropout if num_layers > 1 else 0
        )
        self.text_fc = nn.Linear(hidden_dim * 2, hidden_dim)
        
        # Audio branch (BiLSTM)
        self.layer_norm = nn.LayerNorm(n_mfcc)
        self.audio_bilstm = nn.LSTM(
            input_size=n_mfcc,
            hidden_size=hidden_dim,
            num_layers=num_layers,
            batch_first=True,
            bidirectional=True,
            dropout=dropout if num_layers > 1 else 0
        )
        self.audio_fc = nn.Linear(hidden_dim * 2, hidden_dim)
        
        # Combined
        self.combined_fc = nn.Sequential(
            nn.Linear(hidden_dim * 2, hidden_dim),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(hidden_dim, num_classes)
        )
    
    def forward(self, input_ids, mfcc, seq_lens):
        # Text branch
        embedded = self.embedding(input_ids)  # [batch, max_length, embedding_dim]
        text_out, (hn_text, _) = self.text_bilstm(embedded)  # [batch, max_length, hidden_dim * 2]
        text_features = self.text_fc(torch.cat((hn_text[-2], hn_text[-1]), dim=1))  # [batch, hidden_dim]
        
        # Audio branch
        mfcc = self.layer_norm(mfcc)  # [batch, max_audio_length, n_mfcc]
        packed_audio = nn.utils.rnn.pack_padded_sequence(
            mfcc, seq_lens.cpu(), batch_first=True, enforce_sorted=False
        )
        _, (hn_audio, _) = self.audio_bilstm(packed_audio)  # hn_audio: [num_layers * 2, batch, hidden_dim]
        audio_features = self.audio_fc(torch.cat((hn_audio[-2], hn_audio[-1]), dim=1))  # [batch, hidden_dim]
        
        # Combine
        combined_features = torch.cat((text_features, audio_features), dim=1)  # [batch, hidden_dim * 2]
        output = self.combined_fc(combined_features)  # [batch, num_classes]
        return output

# Train function
from torch.optim.lr_scheduler import ReduceLROnPlateau

def train_model(model, train_loader, val_loader, num_epochs=EPOCHS, learning_rate=LEARNING_RATE, device=DEVICE, patience=3):
    criterion = nn.CrossEntropyLoss()
    optimizer = optim.Adam(model.parameters(), lr=learning_rate)
    scheduler = ReduceLROnPlateau(optimizer, mode='min', factor=0.5, patience=2, min_lr=1e-6, verbose=True)
    best_val_loss = float('inf')
    best_epoch = 0
    patience_counter = 0
    best_model_state = None
    train_losses, val_losses = [], []
    train_f1s, val_f1s = [], []
    
    for epoch in range(num_epochs):
        model.train()
        train_loss = 0.0
        train_preds, train_targets = [], []
        
        for input_ids, mfcc, targets, _, _, _, seq_lens in tqdm(train_loader, desc=f"Training Epoch {epoch + 1}/{num_epochs}"):
            input_ids, mfcc, targets, seq_lens = input_ids.to(device), mfcc.to(device), targets.to(device), seq_lens.to(device)
            
            optimizer.zero_grad()
            outputs = model(input_ids, mfcc, seq_lens)
            loss = criterion(outputs, targets)
            loss.backward()
            optimizer.step()
            
            train_loss += loss.item()
            preds = torch.argmax(outputs, dim=1)
            train_preds.extend(preds.cpu().numpy())
            train_targets.extend(targets.cpu().numpy())
        
        train_loss /= len(train_loader)
        train_wa = accuracy_score(train_targets, train_preds)
        train_ua = f1_score(train_targets, train_preds, average='macro')
        train_f1 = f1_score(train_targets, train_preds, average='weighted')
        train_losses.append(train_loss)
        train_f1s.append(train_f1)
        
        # Validation
        model.eval()
        val_loss = 0.0
        val_preds, val_targets = [], []
        
        with torch.no_grad():
            for input_ids, mfcc, targets, _, _, _, seq_lens in val_loader:
                input_ids, mfcc, targets, seq_lens = input_ids.to(device), mfcc.to(device), targets.to(device), seq_lens.to(device)
                outputs = model(input_ids, mfcc, seq_lens)
                loss = criterion(outputs, targets)
                val_loss += loss.item()
                
                preds = torch.argmax(outputs, dim=1)
                val_preds.extend(preds.cpu().numpy())
                val_targets.extend(targets.cpu().numpy())
        
        val_loss /= len(val_loader)
        val_wa = accuracy_score(val_targets, val_preds)
        val_ua = f1_score(val_targets, val_preds, average='macro')
        val_f1 = f1_score(val_targets, val_preds, average='weighted')
        val_losses.append(val_loss)
        val_f1s.append(val_f1)
        
        current_lr = optimizer.param_groups[0]['lr']
        print(f"Epoch {epoch + 1}/{num_epochs}: "
              f"Train Loss={train_loss:.4f}, Train WA={train_wa:.4f}, Train UA={train_ua:.4f}, Train F1={train_f1:.4f}, "
              f"Val Loss={val_loss:.4f}, Val WA={val_wa:.4f}, Val UA={val_ua:.4f}, Val F1={val_f1:.4f}, "
              f"LR={current_lr:.6f}")
        
        # Early stopping based on validation loss
        if val_loss < best_val_loss:
            best_val_loss = val_loss
            best_epoch = epoch + 1
            best_model_state = model.state_dict()
            patience_counter = 0
            torch.save(best_model_state, MODEL_SAVE_PATH)
            print(f"Saved best model with Val Loss={val_loss:.4f} at {MODEL_SAVE_PATH}")
        else:
            patience_counter += 1
            print(f"No improvement in validation loss. Patience counter: {patience_counter}/{patience}")
            if patience_counter >= patience:
                print(f"Early stopping triggered after {patience_counter} epochs without improvement.")
                break
        
        # Step the scheduler
        scheduler.step(val_loss)
    
    # Restore best model
    if best_model_state is not None:
        model.load_state_dict(best_model_state)
        print(f"Restored best model from epoch {best_epoch} with Val Loss={best_val_loss:.4f}")
    
    # Plot and save loss and F1 curves
    epochs = range(1, len(train_losses) + 1)
    plt.figure(figsize=(10, 5))
    plt.plot(epochs, train_losses, label='Train Loss')
    plt.plot(epochs, val_losses, label='Val Loss')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.title('Training and Validation Loss')
    plt.legend()
    plt.savefig(LOSS_CURVE_PATH)
    plt.close()
    
    plt.figure(figsize=(10, 5))
    plt.plot(epochs, train_f1s, label='Train F1')
    plt.plot(epochs, val_f1s, label='Val F1')
    plt.xlabel('Epoch')
    plt.ylabel('F1 Score')
    plt.title('Training and Validation F1 Score')
    plt.legend()
    plt.savefig(F1_CURVE_PATH)
    plt.close()
    
    print(f"Loss and F1 curves saved to {LOSS_CURVE_PATH} and {F1_CURVE_PATH}")
    return val_f1

# Infer function
def infer(model, test_loader, dataset, utterance_ids=None, max_length=128, emotion_classes=None, device=DEVICE, sample_count=10):
    if emotion_classes is None:
        raise ValueError("emotion_classes must be provided as a list of emotion labels (e.g., ['angry', 'happy', 'sad', 'neutral'])")
    
    model.eval()
    idx_to_emotion = {idx: emo for idx, emo in enumerate(emotion_classes)}
    results = []
    samples_processed = 0
    
    # Determine dataset attributes
    data_source = dataset.dataset.data if isinstance(dataset, torch.utils.data.Subset) else dataset.data
    dataset_dir = dataset.dataset.dataset_dir if isinstance(dataset, torch.utils.data.Subset) else dataset.dataset_dir
    
    with torch.no_grad():
        for batch_idx, (input_ids, mfcc, targets, utt_ids, context_texts, target_texts, seq_lens) in enumerate(tqdm(test_loader, desc=f"Running inference on up to {sample_count} samples")):
            if samples_processed >= sample_count:
                break
            
            # Filter by utterance_ids if provided
            batch_indices = range(len(utt_ids))
            if utterance_ids:
                batch_indices = [i for i in range(len(utt_ids)) if utt_ids[i] in utterance_ids]
                if not batch_indices:
                    continue
            
            input_ids, mfcc, targets, seq_lens = input_ids.to(device), mfcc.to(device), targets.to(device), seq_lens.to(device)
            outputs = model(input_ids, mfcc, seq_lens)
            probs = torch.softmax(outputs, dim=1)
            pred_indices = torch.argmax(probs, dim=1)
            
            for i in batch_indices:
                if samples_processed >= sample_count:
                    break
                
                utt_id = utt_ids[i]
                pred_idx = pred_indices[i].item()
                pred_emotion = idx_to_emotion[pred_idx]
                pred_prob = probs[i, pred_idx].item()
                true_emotion = idx_to_emotion[targets[i].item()]
                
                # Audio file path
                data_idx = test_loader.dataset.indices[batch_idx * test_loader.batch_size + i] if isinstance(test_loader.dataset, torch.utils.data.Subset) else (batch_idx * test_loader.batch_size + i)
                session = data_source.iloc[data_idx]['session']
                dialogue = utt_id.rsplit('_', 1)[0]
                audio_path = os.path.join(dataset_dir, session, 'sentences', 'wav', dialogue, f"{utt_id}.wav")
                audio_exists = os.path.exists(audio_path) if audio_path else False
                
                # Display input text and context
                print(f"\nSample {samples_processed + 1}: utt_id={utt_id}")
                print(f"Target Text: {target_texts[i]}")
                print("Context:")
                for j, context in enumerate(context_texts[i], 1):
                    print(f"  Context {j}: {context}")
                print(f"Predicted: {pred_emotion}, True: {true_emotion}")
                
                # Play audio
                if audio_exists and IPYTHON_AVAILABLE:
                    try:
                        print(f"Playing audio for {utt_id}: {audio_path}")
                        display(Audio(audio_path))
                    except Exception as e:
                        print(f"Failed to play audio for {utt_id}: {e}")
                elif audio_exists:
                    print(f"Audio available but playback unsupported: {audio_path}")
                else:
                    print(f"Audio file missing for {utt_id}: {audio_path}")
                
                # Metadata
                metadata = {
                    'utt_id': utt_id,
                    'audio_path': audio_path,
                    'audio_exists': audio_exists,
                    'predicted_emotion': pred_emotion,
                    'predicted_prob': pred_prob,
                    'true_emotion': true_emotion,
                    'context_texts': context_texts[i],
                    'target_text': target_texts[i]
                }
                results.append(metadata)
                samples_processed += 1
    
    # Save results
    output_df = pd.DataFrame(results)
    output_path = '/kaggle/working/multimodal_test_inference_few_samples.csv'
    output_df.to_csv(output_path, index=False)
    print(f"Inference results for {len(results)} samples saved to {output_path}")
    
    return results

# Evaluate function
def evaluate_model(model, test_loader, emotion_classes):
    model.eval()
    all_preds, all_targets = [], []
    sample_counter = 0
    idx_to_emotion = {idx: emo for idx, emo in enumerate(emotion_classes)}
    
    with torch.no_grad():
        for batch_idx, (input_ids, mfcc, targets, utt_ids, context_texts, target_texts, seq_lens) in enumerate(test_loader):
            # Debug: Print batch shapes
            print(f"Test batch {batch_idx + 1}: input_ids.shape={input_ids.shape}, mfcc.shape={mfcc.shape}, "
                  f"targets.shape={targets.shape}, seq_lens.shape={seq_lens.shape}, utt_ids={utt_ids[:2]}...")
            
            input_ids, mfcc, targets, seq_lens = input_ids.to(DEVICE), mfcc.to(DEVICE), targets.to(DEVICE), seq_lens.to(DEVICE)
            outputs = model(input_ids, mfcc, seq_lens)
            
            # Debug: Print output shape
            print(f"Model output shape: {outputs.shape}")
            
            preds = torch.argmax(outputs, dim=1)
            all_preds.extend(preds.cpu().numpy())
            all_targets.extend(targets.cpu().numpy())
            
            # Display input text, context, and predictions for each sample
            for i in range(len(utt_ids)):
                sample_counter += 1
                print(f"\nSample {sample_counter}: utt_id={utt_ids[i]}")
                print(f"Target Text: {target_texts[i]}")
                print("Context:")
                for j, context in enumerate(context_texts[i], 1):
                    print(f"  Context {j}: {context}")
                print(f"Predicted: {idx_to_emotion[preds[i].item()]}, True: {idx_to_emotion[targets[i].item()]}")
            
    # Compute metrics
    wa = accuracy_score(all_targets, all_preds)
    ua = f1_score(all_targets, all_preds, average='macro')
    f1 = f1_score(all_targets, all_preds, average='weighted')
    
    # Plot confusion matrix
    cm = confusion_matrix(all_targets, all_preds)
    plt.figure(figsize=(8, 6))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=emotion_classes, yticklabels=emotion_classes)
    plt.xlabel('Predicted')
    plt.ylabel('True')
    plt.title('Confusion Matrix')
    plt.savefig('/kaggle/working/multimodal_confusion_matrix.png')
    plt.show()
    
    print(f"Evaluation completed: WA={wa:.4f}, UA={ua:.4f}, F1={f1:.4f}")
    return wa, ua, f1

# Main execution
if __name__ == "__main__":
    # Dataset
    dataset = Multimodal_IEMOCAPDataset(
        csv_path=CSV_PATH,
        dataset_dir=DATASET_DIR,
        mfcc_path=MFCC_PATH,
        sessions=SESSIONS,
        max_length=MAX_LENGTH,
        n_mfcc=N_MFCC,
        max_audio_length=MAX_AUDIO_LENGTH
    )
    print(f"Total dataset size: {len(dataset)}")
    
    # Stratified split
    labels = [dataset.emotion_to_idx[dataset.data.iloc[idx]['mapped_emotion']] for idx in range(len(dataset))]
    indices = np.arange(len(dataset))
    train_indices, temp_indices, train_labels, temp_labels = train_test_split(
        indices,
        labels,
        test_size=0.3,
        stratify=labels,
        random_state=42
    )
    val_indices, test_indices, val_labels, test_labels = train_test_split(
        temp_indices,
        temp_labels,
        test_size=0.5,
        stratify=temp_labels,
        random_state=42
    )
    
    train_dataset = Subset(dataset, train_indices)
    val_dataset = Subset(dataset, val_indices)
    test_dataset = Subset(dataset, test_indices)
    print(f"Train dataset size: {len(train_dataset)}, Val dataset size: {len(val_dataset)}, Test dataset size: {len(test_dataset)}")
    
    # DataLoaders
    train_loader = DataLoader(
        train_dataset,
        batch_size=BATCH_SIZE,
        shuffle=True,
        collate_fn=multimodal_collate_fn,
        num_workers=0
    )
    val_loader = DataLoader(
        val_dataset,
        batch_size=BATCH_SIZE,
        shuffle=False,
        collate_fn=multimodal_collate_fn,
        num_workers=0
    )
    test_loader = DataLoader(
        test_dataset,
        batch_size=BATCH_SIZE,
        shuffle=False,
        collate_fn=multimodal_collate_fn,
        num_workers=0
    )
    print(f"Train DataLoader size: {len(train_loader)}, Val DataLoader size: {len(val_loader)}, Test DataLoader size: {len(test_loader)}")
    
    # Model
    model = Multimodal_LSTMEmotionClassifier(
        vocab_size=len(dataset.word2idx),
        embedding_dim=EMBEDDING_DIM,
        hidden_dim=HIDDEN_DIM,
        num_layers=NUM_LAYERS,
        num_classes=len(EMOTION_CLASSES),
        dropout=DROPOUT,
        embedding_matrix=dataset.embedding_matrix,
        n_mfcc=N_MFCC
    ).to(DEVICE)
    
    # Train
    print("\nTraining model...")
    best_f1 = train_model(model, train_loader, val_loader, num_epochs=EPOCHS, learning_rate=LEARNING_RATE, device=DEVICE)
    print(f"Best validation F1: {best_f1:.4f}")
    
    # Load best model
    model.load_state_dict(torch.load(MODEL_SAVE_PATH))
    model.eval()
    
    # Infer on 10 samples
    print("\nRunning inference on 10 samples from test_loader...")
    test_results = infer(
        model=model,
        test_loader=test_loader,
        dataset=test_dataset,
        utterance_ids=None,
        max_length=MAX_LENGTH,
        emotion_classes=EMOTION_CLASSES,
        device=DEVICE,
        sample_count=10
    )
    for result in test_results:
        print(result)
    
    # Evaluate
    print("\nEvaluating on test_loader...")
    wa, ua, f1 = evaluate_model(model, test_loader, EMOTION_CLASSES)
    print(f"Final metrics: WA={wa:.4f}, UA={ua:.4f}, F1={f1:.4f}")
```
